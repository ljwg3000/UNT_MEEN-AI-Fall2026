# [HW 1-1] Data Indexing
Extract the **sensor signals (excluding the time column)** from **0.02 to 0.12  seconds (not including 0.12 sec.)**  
by indexing the `ExampleData` provided from GitHub.  

* Data link: 'https://github.com/ljwg3000/UNT_MEEN-AI-Fall2026/blob/main/AI_tutorial/DA1/ExampleData?raw=true'
* Index the data as DataFrame format
* Refer to DA1_Code1

```python
# Import the necessary packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load ExampleData from Github
ExampleData = pd.read_csv('https://github.com/ljwg3000/UNT_MEEN-AI-Fall2026/blob/main/AI_tutorial/DA1/ExampleData?raw=true', sep=',', header=None)

# Data indexing
StartPoint = np.where(ExampleData.iloc[:,0].values == 0.02)[0][0]
EndPoint   = np.where(ExampleData.iloc[:,0].values == 0.12)[0][0]


# Define the indexed data as 'SensorData'
SensorData = ExampleData.iloc[StartPoint:EndPoint, 1:]

SensorData
```

# [HW 1-2] Data Processing & Plotting
For each sensor signal in `SensorData`,  
calculate the **mean value every 0.01 seconds** and plot how it changes over time.  

* Refer to DA1_Code3

```python
# Import additional packages in needed


# Feature extraction
Mean_Values = np.zeros((10 , SensorData.shape[1]))

for sensor in range(SensorData.shape[1]):
    for time in range(10):
        Mean_Values[time,sensor] = np.mean(SensorData.iloc[128*(time):128*(time+1), sensor])

print(Mean_Values)

# Plot graphs
TimeArr = np.arange(2,12)/100

plt.figure(figsize=(10,5))
plt.plot(TimeArr, Mean_Values[:,0], ls = '--', c = 'r', marker = 'o', ms = 5, mfc = 'r', mec = 'r') # Acceleration
plt.plot(TimeArr, Mean_Values[:,1], ls = '--', c = 'g', marker = 'o', ms = 5, mfc = 'g', mec = 'g') # Voltage
plt.plot(TimeArr, Mean_Values[:,2], ls = '--', c = 'b', marker = 'o', ms = 5, mfc = 'b', mec = 'b') # Current
plt.xlabel('time(s)')
plt.ylabel('Mean value')
plt.legend(['Mean_Acc','Mean_Vol', 'Mean_Cur'])
plt.show()
```
